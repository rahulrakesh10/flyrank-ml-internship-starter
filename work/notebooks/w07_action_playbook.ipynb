## 1. Ranked actions + reason codes

*The queue: what to do first, and why, in words a human trusts.*

**The Action Queue**: This playbook prioritizes content that is actively failing to capture existing search demand despite having potential. We rank items using a combination of Search Volume and our baseline rules, validated against directional ML analysis.

**Reason Codes**:
- `top_10_low_ctr`: *"Ranking on page 1 but failing to get clicks."* Action: Rewrite Meta Title & Description to improve relevance and click-appeal.
- `striking_distance`: *"Ranking 11-20 with high volume."* Action: Expand word count or add internal links to push onto page 1.


```python
import pandas as pd
import os

df = pd.read_csv('../../data/raw/content_refresh_anonymized.csv')

def get_action_reason(row):
    if row['search_volume'] < 100 or row['avg_position'] == 0:
        return None
    if row['avg_position'] <= 10 and row['ctr'] < 1.0:
        return 'top_10_low_ctr'
    elif 10 < row['avg_position'] <= 20:
        return 'striking_distance'
    return None

df['reason_code'] = df.apply(get_action_reason, axis=1)
playbook_df = df[df['reason_code'].notnull()].copy()
playbook_df['priority_score'] = playbook_df['search_volume']
playbook_df = playbook_df.sort_values('priority_score', ascending=False)

print(f"Total items requiring action: {len(playbook_df)}")
print("Top 5 actions:")
display(playbook_df[['content_id', 'reason_code', 'search_volume', 'avg_position']].head())
```

## 3. Human review + the no-go list

*What a person must check before acting. What should never be automated.*

**Human Review Checklist before Execution**:
1. Check the Search Engine Results Page (SERP) manually: Is the search intent informational or navigational?
